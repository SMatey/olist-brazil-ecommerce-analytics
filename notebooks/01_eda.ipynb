### Verificar valores nulos, registros faltantes y duplicados.

```python
# Verificar valores nulos y duplicados del DataFrame orders
print("=" * 70)
print("VALORES NULOS Y DUPLICADOS DEL DATAFRAME: orders")
print("=" * 70)

# Valores nulos por columna
null_values_orders = orders.isnull().sum()

# Verificar duplicados
duplicates_orders = orders.duplicated().sum()

# Mostrar resultados
print("\nValores nulos por columna:\n", null_values_orders)
print("\nNúmero de registros duplicados:", duplicates_orders)
print("\nTotal de registros:", len(orders))
print("Porcentaje de duplicados: {:.2f}%".format((duplicates_orders / len(orders)) * 100))
```

### Verificar valores extremos (outliers).

```python
# Verificar outliers del DataFrame order_items
print("=" * 70)
print("OUTLIERS DEL DATAFRAME: order_items")
print("=" * 70)

# Seleccionar solo columnas numéricas
numeric_columns_order_items = order_items.select_dtypes(include=[np.number])

if not numeric_columns_order_items.empty:
    for column in numeric_columns_order_items.columns:
        # Calcular Z-score
        z_scores = (numeric_columns_order_items[column] - numeric_columns_order_items[column].mean()) / numeric_columns_order_items[column].std()
        
        # Identificar outliers (Z-score > 3 o < -3)
        outliers = abs(z_scores) > 3
        num_outliers = outliers.sum()
        
        print(f"\nColumna: {column}")
        print(f"Número de outliers: {num_outliers}")
        print(f"Porcentaje de outliers: {(num_outliers / len(numeric_columns_order_items)) * 100:.2f}%")
        
        if num_outliers > 0:
            print(f"Valores outliers detectados:\n{numeric_columns_order_items[outliers][column].head(10)}")
else:
    print("\nNo hay columnas numéricas en este DataFrame.")
```

En order_items aparecen pocos casos extremos en las tres columnas analizadas, con 1.68% en order_item_id, 1.75% en price esto indica importes muy altos respecto al precio típico, y 1.81% en freight_value indicando fletes elevados. Estos resultados confirman una cola larga, ya que la mayoría de filas concentra valores bajos o moderados, pero existe un grupo pequeño con precios y fletes muy altos y pedidos con varios ítems que empujan las medidas basadas en promedios.
