## Limpieza de datos, archivo: Orders reviews

```python
import pandas as pd

RAW = "../../data/raw/"

# =========================
# Carga de datos
# =========================
order_reviews = pd.read_csv(
    RAW + "olist_order_reviews_dataset.csv",
    parse_dates=["review_creation_date", "review_answer_timestamp"]
)
```

```python
order_reviews = order_reviews.drop_duplicates(subset=["review_id"], keep="first").copy()

# =========================
# Rellenar con etiquetas 
# =========================

# Banderas (1 si originalmente era nulo, 0 si no)
order_reviews["had_title_na"]   = order_reviews["review_comment_title"].isna().astype(int)
order_reviews["had_message_na"] = order_reviews["review_comment_message"].isna().astype(int)

# Relleno con textos neutros
order_reviews["review_comment_title"]   = order_reviews["review_comment_title"].fillna("Unknown")
order_reviews["review_comment_message"] = order_reviews["review_comment_message"].fillna("No_comment")

# =========================
# Resultado
# =========================
order_reviews_clean = order_reviews.copy()

print({
    "rows_final": len(order_reviews_clean),
    "filled_titles": int(order_reviews_clean["had_title_na"].sum()),
    "filled_messages": int(order_reviews_clean["had_message_na"].sum())
})

order_reviews_clean.head()
```
