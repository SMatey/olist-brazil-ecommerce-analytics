## Limpieza de datos, archivo: Orders reviews

```python
# === Bloque 1: Limpieza =======================================================
import sys, os
PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..', '..')) 
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

import pandas as pd
from src.olistprep.data_cleaning import remove_duplicates, fill_missing_values

# Cargar el DataFrame
order_reviews_df = pd.read_csv('../../data/raw/olist_order_reviews_dataset.csv')

# Tamaño inicial
shape_before = order_reviews_df.shape

# Eliminar duplicados (EDA: 0; llamada inocua. Se usa fecha de creación para priorizar el más reciente si existiera)
order_reviews_df_clean = remove_duplicates(order_reviews_df, 'review_creation_date')

# Rellenar valores nulos (imputa 'Unknown' en columnas de texto si hubiera)
order_reviews_df_clean = fill_missing_values(order_reviews_df_clean)

# Tamaño final
shape_after = order_reviews_df_clean.shape
```

## Resultados de limpieza

```python
# === Bloque 2: Resultados =====================================================
# Duplicados residuales (registro completo)
duplicates_after = order_reviews_df_clean.duplicated().sum()

# Nulos por columna después de la limpieza / (o del drop, si lo activaste)
nulls_after = order_reviews_df_clean.isnull().sum()

print("=== Resumen de limpieza: order_reviews ===")
print(f"Tamaño inicial: {shape_before}")
print(f"Tamaño después de limpieza: {shape_after}")
print(f"Duplicados después (registro completo): {duplicates_after}")
print("\nNulos por columna (post-limpieza):")
print(nulls_after)

print("\nPrimeras filas después de la limpieza:")
display(order_reviews_df_clean.head())
```

## Guardar DataFrame limpio 

```python
# === Bloque 3: Guardado =======================================================
save_path = '../../data/processed/olist_order_reviews_clean.csv'
order_reviews_df_clean.to_csv(save_path, index=False)
print(f"Archivo guardado en: {save_path}")
```
