## Limpieza de datos, archivo: Sellers

```python
# === Bloque 1: Limpieza =======================================================
import sys, os
PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..', '..')) 
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

import pandas as pd
from src.olistprep.data_cleaning import remove_duplicates, fill_missing_values

# Cargar el DataFrame
sellers_df = pd.read_csv('../../data/raw/olist_sellers_dataset.csv')

# Tamaño inicial
shape_before = sellers_df.shape

# NO deduplicar: EDA indica 0 duplicados 
sellers_df_clean = sellers_df.copy()

# Rellenar valores nulos (EDA: no hay nulos, se mantiene por consistencia, no debe cambiar nada)
sellers_df_clean = fill_missing_values(sellers_df_clean)

# (Sin outliers: NO usar remove_outliers)

# Tamaño final
shape_after = sellers_df_clean.shape
```

## Resultados de limpieza

```python
# === Bloque 2: Resultados =====================================================
duplicates_after = sellers_df_clean.duplicated().sum()
nulls_after = sellers_df_clean.isnull().sum()

print("=== Resumen de limpieza: sellers ===")
print(f"Tamaño inicial: {shape_before}")
print(f"Tamaño después de limpieza: {shape_after}")
print(f"Duplicados después (registro completo): {duplicates_after}")
print("\nNulos por columna (post-limpieza):")
print(nulls_after)

print("\nPrimeras filas después de la limpieza:")
display(sellers_df_clean.head())
```

## Guardar DataFrame limpio 

```python
# === Bloque 3: Guardado =======================================================
save_path = '../../data/processed/olist_sellers_clean.csv'
sellers_df_clean.to_csv(save_path, index=False)
print(f"Archivo guardado en: {save_path}")
```
