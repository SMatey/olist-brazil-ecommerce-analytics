```python
# =========================
# Importaciones
# =========================
import pandas as pd

# Ruta base (ajústala si corresponde)
RAW = "../../data/raw/"

# =========================
# Carga de datos
# =========================
sellers = pd.read_csv(RAW + "olist_sellers_dataset.csv")
print(f"Filas iniciales: {len(sellers):,}")

# =========================
# Limpieza mínima según EDA
# =========================

# 1) Duplicados (EDA: 0, dejamos por robustez)
sellers = sellers.drop_duplicates(subset=["seller_id"], keep="first").copy()

# 2) Nulos: no hay -> sin acción
# 3) Timestamps: no hay -> sin acción
# 4) Outliers: no hay en seller_zip_code_prefix -> sin acción

# =========================
# Resultado final
# =========================
sellers_clean = sellers.copy()
print(f"Filas finales: {len(sellers_clean):,}")
print(sellers_clean.dtypes)

# Vista rápida
sellers_clean.head()
```
