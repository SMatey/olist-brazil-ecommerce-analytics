## Limpieza de datos, archivo: Orders payments

```python
import pandas as pd
import numpy as np

RAW = "../../data/raw/"

# =========================
# Carga de datos
# =========================
order_payments = pd.read_csv(RAW + "olist_order_payments_dataset.csv")

print(f"Filas iniciales: {len(order_payments):,}")
```

```python
# =========================
# Limpieza mínima según el EDA
# - Outliers ±3σ en métricas numéricas
# =========================

numeric_metric_cols = ["payment_installments", "payment_value"] 

def remove_outliers_3sigma(df, cols):
    """Devuelve df filtrado a ±3 desviaciones estándar por cada columna en cols."""
    if not cols:
        return df
    mask = pd.Series(True, index=df.index)
    for c in cols:
        x = df[c].astype(float)
        mu = x.mean()
        sigma = x.std(ddof=0)
        # Si sigma=0 o NaN, no filtramos por esa columna
        if np.isnan(mu) or np.isnan(sigma) or sigma == 0:
            continue
        mask &= (x >= mu - 3*sigma) & (x <= mu + 3*sigma)
    return df[mask].copy()

before = len(order_payments)
payments_clean = remove_outliers_3sigma(order_payments, numeric_metric_cols)
removed = before - len(payments_clean)

print(f"Eliminadas por outliers (±3σ) en {numeric_metric_cols}: {removed:,}")
print(f"Filas finales: {len(payments_clean):,}")
payments_clean.head()
```
