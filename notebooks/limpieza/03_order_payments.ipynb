## Limpieza de datos, archivo: Orders payments

```python
# === Bloque 1: Limpieza ====
import sys, os
PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..', '..')) 
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

import pandas as pd
from src.olistprep.data_cleaning import remove_duplicates, fill_missing_values

# Cargar el DataFrame
order_payments_df = pd.read_csv('../../data/raw/olist_order_payments_dataset.csv')

# Tamaño inicial
shape_before = order_payments_df.shape

# NO deduplicar: varias transacciones por orden son válidas (payment_sequential)
order_payments_df_clean = order_payments_df.copy()

# Rellenar valores nulos (EDA: no hay, se mantiene por consistencia)
order_payments_df_clean = fill_missing_values(order_payments_df_clean)

# NO eliminar outliers: colas altas de cuotas/valores son propias del dominio

# Tamaño final
shape_after = order_payments_df_clean.shape
```

## Resultados de limpieza

```python
# === Bloque 2: Resultados ===
# Duplicados residuales (registro completo)
duplicates_after = order_payments_df_clean.duplicated().sum()

# Nulos por columna después de la limpieza
nulls_after = order_payments_df_clean.isnull().sum()

print("=== Resumen de limpieza: order_payments ===")
print(f"Tamaño inicial: {shape_before}")
print(f"Tamaño después de limpieza: {shape_after}")
print(f"Duplicados después (registro completo): {duplicates_after}")
print("\nNulos por columna (post-limpieza):")
print(nulls_after)

print("\nPrimeras filas después de la limpieza:")
display(order_payments_df_clean.head())
```

## Guardar DataFrame limpio 

```python
# === Bloque 3: Guardado ====
save_path = '../../data/processed/olist_order_payments_clean.csv'
order_payments_df_clean.to_csv(save_path, index=False)
print(f"Archivo guardado en: {save_path}")
```
