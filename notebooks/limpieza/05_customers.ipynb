```python
# =========================
# Importaciones
# =========================
import pandas as pd

# Ruta base (ajústala si corresponde)
RAW = "../../data/raw/"

# =========================
# Carga de datos
# =========================
customers = pd.read_csv(RAW + "olist_customers_dataset.csv")

print(f"Filas iniciales: {len(customers):,}")

# =========================
# Limpieza mínima según EDA
# =========================

# Duplicados 
customers = customers.drop_duplicates(subset=["customer_id"], keep="first").copy()

# =========================
# Resultado final
# =========================
customers_clean = customers.copy()
print(f"Filas finales: {len(customers_clean):,}")
print(customers_clean.dtypes)

# Vista rápida
customers_clean.head()
```
