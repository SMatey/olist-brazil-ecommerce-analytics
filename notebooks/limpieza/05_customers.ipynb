## Limpieza de datos, archivo: Customers

```python
# === Bloque 1: Limpieza =======================================================
import sys, os
PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..', '..')) 
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

import pandas as pd
from src.olistprep.data_cleaning import remove_duplicates, fill_missing_values

# Cargar el DataFrame
customers_df = pd.read_csv('../../data/raw/olist_customers_dataset.csv')

# Tamaño inicial
shape_before = customers_df.shape

# NO deduplicar: EDA indica 0 duplicados y no existe columna de fecha ni 'order_id' en este DF.
customers_df_clean = customers_df.copy()

# Rellenar valores nulos (EDA: no hay nulos; se mantiene por consistencia, no debe cambiar nada)
customers_df_clean = fill_missing_values(customers_df_clean)

# (Sin outliers: NO usar remove_outliers)

# Tamaño final
shape_after = customers_df_clean.shape
```

## Resultados de limpieza

```python
# === Bloque 2: Resultados =====================================================
# Duplicados residuales (registro completo)
duplicates_after = customers_df_clean.duplicated().sum()

# Nulos por columna después de la limpieza
nulls_after = customers_df_clean.isnull().sum()

print("=== Resumen de limpieza: customers ===")
print(f"Tamaño inicial: {shape_before}")
print(f"Tamaño después de limpieza: {shape_after}")
print(f"Duplicados después (registro completo): {duplicates_after}")
print("\nNulos por columna (post-limpieza):")
print(nulls_after)

print("\nPrimeras filas después de la limpieza:")
display(customers_df_clean.head())
```
