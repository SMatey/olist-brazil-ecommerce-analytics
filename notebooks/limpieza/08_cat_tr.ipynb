```python
# =========================
# Importaciones
# =========================
import pandas as pd

# Ruta base (ajústala si corresponde)
RAW = "../../data/raw/"

# =========================
# Carga de datos
# =========================
cat_tr = pd.read_csv(RAW + "product_category_name_translation.csv")

print(f"Filas iniciales: {len(cat_tr):,}")

# =========================
# Limpieza mínima según EDA
# =========================

# 1) Duplicados (EDA: 0; línea por robustez)
cat_tr = cat_tr.drop_duplicates(
    subset=["product_category_name", "product_category_name_english"],
    keep="first"
).copy()

# 2) Nulos: no hay -> sin acción
# 3) Timestamps/numéricas/outliers: no aplica (solo categóricas)

# =========================
# Resultado final
# =========================
cat_tr_clean = cat_tr.copy()
print(f"Filas finales: {len(cat_tr_clean):,}")
print(cat_tr_clean.dtypes)

# Vista rápida
cat_tr_clean.head()
```
