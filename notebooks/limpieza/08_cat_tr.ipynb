## Limpieza de datos, archivo: Category_name_translation (cat_tr)

```python
# === Bloque 1: Limpieza =======================================================
import sys, os
PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), '..', '..')) 
if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

import pandas as pd
from src.olistprep.data_cleaning import remove_duplicates, fill_missing_values

# Cargar el DataFrame
cat_trans_df = pd.read_csv('../../data/raw/product_category_name_translation.csv')

# Tamaño inicial
shape_before = cat_trans_df.shape

# NO deduplicar: EDA indica 0 duplicados. Preservar mapeo 1:1 de categorías.
cat_trans_df_clean = cat_trans_df.copy()

# Rellenar valores nulos (EDA: no hay nulos; se mantiene por consistencia, no debe cambiar nada)
cat_trans_df_clean = fill_missing_values(cat_trans_df_clean)

# (Sin outliers: NO usar remove_outliers)

# Tamaño final
shape_after = cat_trans_df_clean.shape
```

## Resultados de limpieza

```python
# === Bloque 2: Resultados =====================================================
duplicates_after = cat_trans_df_clean.duplicated().sum()
nulls_after = cat_trans_df_clean.isnull().sum()

print("=== Resumen de limpieza: product_category_name_translation ===")
print(f"Tamaño inicial: {shape_before}")
print(f"Tamaño después de limpieza: {shape_after}")
print(f"Duplicados después (registro completo): {duplicates_after}")
print("\nNulos por columna (post-limpieza):")
print(nulls_after)

print("\nPrimeras filas después de la limpieza:")
display(cat_trans_df_clean.head())
```
